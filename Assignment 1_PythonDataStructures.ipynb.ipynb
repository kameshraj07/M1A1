**7) Lambda Filtering & List Comprehension**

Use a lambda function and filter() (or a list comprehension with a lambda-like expression) to
extract job roles where the skill count (from skill_count) for any skill >= 2 — that is, roles that
include at least one relatively common skill.

```python
from collections import Counter
skill_count = Counter()


job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": [
        "Data Analyst",
        "Data Scientist",
        "Business Analyst",
        "Data Engineer",
        "ML Engineer"
    ],
    "Company": [
        "ABC Corp",
        "XyZ Ltd",
        "DataWorks",
        "InfraTech",
        "AI Labs"
    ],
    "Skills": [
        "  Excel,SQL,Python   ",
        "  Python, ML, Statistics   ",
        "Excel, SQL, PowerBI",
        "Python ,SQL ,Spark",
        "Python ,ML ,DeepLearning"
    ]
}

for skills in job_data["Skills"]:
  skill_list = [skill.strip() for skill in skills.split(",")]
  skill_count.update(skill_list)

print(skill_list)
print(skill_count)


roles_common_skill = list(filter(
    lambda role_skills: any(skill_count[skill] >= 2 for skill in role_skills[1]),
    zip( job_data["Role"],[[skill.strip() for skill in skills.split(",")]
         for skills in job_data["Skills"]]
    )
))


roles_common_skill = [role for role, skills in roles_common_skill]

print("Roles with at least one common skill:", roles_common_skill)
```

**9) Save Cleaned Data to File (file handling & exceptions)**

Save the cleaned & structured job records to a file named job_skills.csv. Each line should contain:

Job_ID,Role, Company, Skills

1,Data Analyst,ABC Corp,excel sql python

```python
import csv

job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": [
        "Data Analyst",
        "Data Scientist",
        "Business Analyst",
        "Data Engineer",
        "ML Engineer"
    ],
    "Company": [
        "ABC Corp",
        "XyZ Ltd",
        "DataWorks",
        "InfraTech",
        "AI Labs"
    ],
    "Skills": [
        "  Excel,SQL,Python   ",
        "  Python, ML, Statistics   ",
        "Excel, SQL, PowerBI",
        "Python ,SQL ,Spark",
        "Python ,ML ,DeepLearning"
    ]
}

try:
    with open("job_skills.csv", "w", newline="") as file:
        writer = csv.writer(file)

        # Write header
        writer.writerow(["Job_ID", "Role", "Company", "Skills"])

        # Write cleaned job records
        for i in range(len(job_data["Job_ID"])):
            job_id = job_data["Job_ID"][i]
            role = job_data["Role"][i].strip()
            company = job_data["Company"][i].strip()

            # Clean and format skills
            skills = " ".join(
                skill.strip().lower()
                for skill in job_data["Skills"][i].split(",")
            )

            writer.writerow([job_id, role, company, skills])

    print("Cleared data successfully to job_skills.csv")

except FileNotFoundError:
    print("Error: File or directory not found.")

except Exception as e:
    print("An unexpected error occurred:", e)
```
